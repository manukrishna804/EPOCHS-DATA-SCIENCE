```python
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

print(num_cols)
print(cat_cols)
```

```python
corr = df[num_cols].corr()

plt.figure(figsize=(12,10))
sns.heatmap(corr,cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    y="track_genre",
    order=df["track_genre"].value_counts().head(10).index
)

plt.title("Top 10 Genres")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["popularity"],bins=30,kde=True)

plt.title("Popularity Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="danceability",
    y="energy",
    alpha=0.6
)

plt.title("Danceability vs Energy")
plt.show()
```

```python
top=df.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top.values,
    y=top.index
)

plt.title("Average Popularity by Genre")
plt.show()
```

```python
plt.figure(figsize=(6,6))

df["explicit"].value_counts().plot.pie(
    autopct="%1.1f%%",
    startangle=90
)

plt.ylabel("")
plt.title("Explicit vs Non-Explicit Songs")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["duration_ms"])

plt.title("Distribution of Track Duration")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="tempo",
    y="popularity"
)

plt.title("Tempo vs Popularity")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="explicit",
    y="popularity"
)
```
